```python
asia=df[df['origin']=='Asia']
```

```python
asia=asia[['vehicle_weight','model_year']].head(7)
asia
```

```python
X=np.zeros((7,2))
```

```python
X[:,0]=asia['vehicle_weight']
X[:,1]=asia['model_year']
X
```

```python
XT=X.T
```

```python
XTX_inv=np.linalg.inv((XT).dot(X))
XTX_inv
```

```python
y=[1100, 1300, 800, 900, 1000, 1100, 1200]
y
```

```python
w=(XTX_inv.dot(XT)).dot(y)
w
```
